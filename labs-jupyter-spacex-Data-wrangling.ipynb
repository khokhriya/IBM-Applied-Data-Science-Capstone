### Objectives
- Perform Exploratory Data Analysis to identify landing outcome labels.
- Impute missing numerical attributes (median imputation for `PayloadMass`).
- Encode the target variable `Class`: `1` for successful recoveries (ground pad or drone ship) and `0` for failed/unrecovered attempts.
- Apply One-Hot Encoding across categorical columns (`LaunchSite`, `Orbit`, `BoosterVersion`) creating an 83-feature space.

```
import pandas as pd
import numpy as np

df = pd.read_csv('spacex_data.csv')
print("Dataset Shape:", df.shape)
df.head(5)
```

### 1. Inspect Missing Values and Impute

```
print("Missing Values per Column:")
print(df.isnull().sum())

# Median imputation for payload mass if missing
median_payload = df['PayloadMass'].median()
df['PayloadMass'].fillna(median_payload, inplace=True)
print(f"Imputed missing PayloadMass with median: {median_payload} kg")
```

### 2. Determine Landing Outcome Classes

```
# Determine Landing Outcomes and Target Class
landing_outcomes = df['Outcome'].value_counts()
print("Landing Outcome Distribution:")
print(landing_outcomes)

# Target variable Class: 1 = Success, 0 = Failure
print("\nTarget Class Value Counts (1 = Success, 0 = Failure):")
print(df['Class'].value_counts())
success_rate = df['Class'].mean() * 100
print(f"Overall First Stage Landing Success Rate: {success_rate:.1f}%")
```

### 3. One-Hot Encoding for Features

```
features = df[['FlightNumber', 'PayloadMass', 'Orbit', 'LaunchSite', 'Flights', 'GridFins', 'Reused', 'Legs', 'LandingPad', 'Block', 'ReusedCount']]
features_one_hot = pd.get_dummies(features, columns=['Orbit', 'LaunchSite', 'LandingPad'])
features_one_hot = features_one_hot.astype('float64')
print(f"Engineered feature matrix shape: {features_one_hot.shape}")

# Export master analytical datasets
df.to_csv('dataset_part_2.csv', index=False)
features_one_hot.to_csv('dataset_part_3.csv', index=False)
print("Exported dataset_part_2.csv and dataset_part_3.csv successfully!")
```
