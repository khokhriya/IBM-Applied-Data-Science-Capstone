### Objectives
- Visualize flight numbers against launch sites to observe experience learning curves.
- Analyze payload mass distributions per launch facility.
- Evaluate landing success rates categorized by orbital destination (LEO, GTO, SSO, etc.).
- Plot multi-variable scatter plots comparing flight number and payload mass with orbit types.
- Trace the historical year-over-year booster recovery success progression (2010–2020).

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
df = pd.read_csv('spacex_data.csv')
print("Loaded analytical dataset with", len(df), "Falcon 9 records.")
```

### Chart 1: Flight Number vs. Launch Site

```
plt.figure(figsize=(10, 6))
sns.scatterplot(x='FlightNumber', y='LaunchSite', hue='Class', data=df, palette=['#DC2626', '#16A34A'], s=100)
plt.title('Flight Number vs. Launch Site by Landing Outcome', fontsize=14, fontweight='bold')
plt.xlabel('Flight Number', fontsize=12)
plt.ylabel('Launch Site', fontsize=12)
plt.show()
```

### Chart 2: Payload Mass vs. Launch Site

```
plt.figure(figsize=(10, 6))
sns.scatterplot(x='PayloadMass', y='LaunchSite', hue='Class', data=df, palette=['#DC2626', '#16A34A'], s=100)
plt.title('Payload Mass (kg) vs. Launch Site', fontsize=14, fontweight='bold')
plt.xlabel('Payload Mass (kg)', fontsize=12)
plt.ylabel('Launch Site', fontsize=12)
plt.show()
```

### Chart 3: Landing Success Rate by Orbit Type

```
orbit_success = df.groupby('Orbit')['Class'].mean().reset_index()
plt.figure(figsize=(10, 6))
sns.barplot(x='Orbit', y='Class', data=orbit_success, palette='Blues_d')
plt.title('Landing Success Rate by Orbit Type', fontsize=14, fontweight='bold')
plt.xlabel('Orbit Type', fontsize=12)
plt.ylabel('Success Rate', fontsize=12)
plt.ylim(0, 1.05)
plt.show()
```

### Chart 4: Flight Number vs. Orbit Type

```
plt.figure(figsize=(10, 6))
sns.scatterplot(x='FlightNumber', y='Orbit', hue='Class', data=df, palette=['#DC2626', '#16A34A'], s=100)
plt.title('Flight Number vs. Orbit Type', fontsize=14, fontweight='bold')
plt.xlabel('Flight Number', fontsize=12)
plt.ylabel('Orbit Type', fontsize=12)
plt.show()
```

### Chart 5: Payload Mass vs. Orbit Type

```
plt.figure(figsize=(10, 6))
sns.scatterplot(x='PayloadMass', y='Orbit', hue='Class', data=df, palette=['#DC2626', '#16A34A'], s=100)
plt.title('Payload Mass vs. Orbit Type', fontsize=14, fontweight='bold')
plt.xlabel('Payload Mass (kg)', fontsize=12)
plt.ylabel('Orbit Type', fontsize=12)
plt.show()
```

### Chart 6: Yearly Average Landing Success Trend

```
df['Year'] = pd.to_datetime(df['Date']).dt.year
yearly_success = df.groupby('Year')['Class'].mean().reset_index()
plt.figure(figsize=(10, 6))
sns.lineplot(x='Year', y='Class', data=yearly_success, marker='o', color='#0B49CB', linewidth=2.5)
plt.title('SpaceX Falcon 9 First Stage Recovery Yearly Success Trend (2010 - 2020)', fontsize=14, fontweight='bold')
plt.xlabel('Year', fontsize=12)
plt.ylabel('Average Success Rate', fontsize=12)
plt.ylim(0, 1.05)
plt.show()
```
