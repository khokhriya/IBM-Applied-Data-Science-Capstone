### Objectives
- Create an embedded SQLite database instance (`spacex.db`).
- Ingest historical launch table `SPACEXTBL`.
- Execute 10 required Capstone SQL queries to analyze launch pad frequency, customer payload totals, booster variants, and outcome rankings.

```
import sqlite3
import pandas as pd

# Initialize SQLite database and populate table
conn = sqlite3.connect('spacex.db')
df = pd.read_csv('spacex_data.csv')
df.to_sql('SPACEXTBL', conn, if_exists='replace', index=False)
print("Loaded SPACEXTBL into SQLite database with", len(df), "records.")
```

### Query 1: Find Unique Launch Sites

```
q1 = pd.read_sql_query("SELECT DISTINCT LaunchSite FROM SPACEXTBL;", conn)
print(q1)
```

### Query 2: Launch Sites Beginning with 'CCA'

```
q2 = pd.read_sql_query("SELECT FlightNumber, LaunchSite, Date, PayloadMass FROM SPACEXTBL WHERE LaunchSite LIKE 'CCA%' LIMIT 5;", conn)
print(q2)
```

### Query 3: Total NASA Payload Mass

```
q3 = pd.read_sql_query("SELECT SUM(PayloadMass) AS Total_NASA_Payload_KG FROM SPACEXTBL WHERE Orbit = 'ISS';", conn)
print(q3)
```

### Query 4: Average Payload Mass for Falcon 9 v1.1

```
q4 = pd.read_sql_query("SELECT AVG(PayloadMass) AS Avg_Payload_KG FROM SPACEXTBL WHERE BoosterVersion LIKE '%v1.1%';", conn)
print(q4)
```

### Query 5: First Successful Ground Landing Date

```
q5 = pd.read_sql_query("SELECT MIN(Date) AS First_Ground_Landing, BoosterVersion, LaunchSite FROM SPACEXTBL WHERE Class = 1 AND Outcome LIKE '%RTLS%';", conn)
print(q5)
```

### Query 6: Successful Drone Ship Landings with Payload Between 4,000 and 6,000 kg

```
q6 = pd.read_sql_query("SELECT BoosterVersion, LaunchSite, PayloadMass FROM SPACEXTBL WHERE Class = 1 AND Outcome LIKE '%ASDS%' AND PayloadMass BETWEEN 4000 AND 6000;", conn)
print(q6)
```

### Query 7: Total Number of Successful and Failure Mission Outcomes

```
q7 = pd.read_sql_query("SELECT Class, COUNT(*) AS Total_Count FROM SPACEXTBL GROUP BY Class;", conn)
print(q7)
```

### Query 8: Boosters Carrying Maximum Payload

```
q8 = pd.read_sql_query("SELECT BoosterVersion, PayloadMass FROM SPACEXTBL WHERE PayloadMass = (SELECT MAX(PayloadMass) FROM SPACEXTBL);", conn)
print(q8)
```

### Query 9: 2015 Drone Ship Landing Failure Analysis

```
q9 = pd.read_sql_query("SELECT Date, BoosterVersion, LaunchSite, Outcome FROM SPACEXTBL WHERE strftime('%Y', Date) = '2015' AND Outcome LIKE '%False ASDS%';", conn)
print(q9)
```

### Query 10: Rank Landing Outcomes Between 2010-06-04 and 2017-03-20

```
q10 = pd.read_sql_query("SELECT Outcome, COUNT(*) AS Outcome_Count FROM SPACEXTBL WHERE Date BETWEEN '2010-06-04' AND '2017-03-20' GROUP BY Outcome ORDER BY Outcome_Count DESC;", conn)
print(q10)
```
