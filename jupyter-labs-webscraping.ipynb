### Objectives
- Web scrape Falcon 9 launch records from Wikipedia using BeautifulSoup4.
- Parse HTML table structures, targeting flight numbers, launch dates, booster versions, launch sites, payloads, and landing outcomes.
- Sanitize footnote references (e.g. `[12]`) and extract structured data into a pandas DataFrame.
- Validate web scraped records against official API launch logs.

```
import sys
import requests
from bs4 import BeautifulSoup
import re
import unicodedata
import pandas as pd

print("BeautifulSoup and scraping libraries loaded successfully!")
```

### 1. Request Wikipedia Launch Page

```
static_url = "https://en.wikipedia.org/w/index.php?title=List_of_Falcon_9_and_Falcon_Heavy_launches&oldid=1027686922"
response = requests.get(static_url)
print("Wikipedia HTTP Status:", response.status_code)
soup = BeautifulSoup(response.text, 'html.parser')
html_tables = soup.find_all('table')
print(f"Total HTML tables identified: {len(html_tables)}")
```
